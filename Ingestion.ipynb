```python
import pandas as pd
import os
from sqlalchemy import create_engine
```

```python
engine = create_engine('sqlite:///inventory.db')
```

```python
def ingestion(dataframe, table_name, engine):
    dataframe.to_sql(table_name, con = engine)
    
path = 'D:/Data sets/Hotel_dataset/'
for file in os.listdir(path):
    print(file)
    if '.csv' in file:
        df = pd.read_csv('D:/Data sets/Hotel_dataset/hotel_bookings 2.csv')

        ingestion(df,file[:-4], engine)       
```
